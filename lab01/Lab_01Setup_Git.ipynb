```python
import sys
import platform
import os
import importlib.metadata as md

print("=== Inspection 1: Python and system ===")
print(f"Python  : {sys.version.split()[0]}")
print(f"OS      : {platform.system()} {platform.release()}")
print(f"CPU cores: {os.cpu_count()}")

print("\n=== Inspection 2: Key libraries ===")
packages = [
    "numpy",
    "pandas",
    "requests",
    "beautifulsoup4",
    "spacy",
    "matplotlib",
    "seaborn",
    "textblob",
    "scikit-learn",
]

for pkg in packages:
    try:
        ver = md.version(pkg)
    except Exception:
        ver = "Not installed"
    print(f"{pkg:<14} {ver}")
```
